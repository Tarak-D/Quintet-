```python
train = train.merge(
    question_meta[
        ["QuestionId", "SubjectId"]
    ],
    on="QuestionId",
    how="left"
)

print("Shape after merge:", train.shape)

print("\nMissing SubjectIds:")
print(train["SubjectId"].isna().sum())

print("\nSample:")
display(train.head())
```

```python
overall_accuracy = train["IsCorrect"].mean()

student_stats = (
    train.groupby("UserId")
    .agg(
        total_attempts=("IsCorrect", "count"),
        correct_answers=("IsCorrect", "sum"),
        accuracy=("IsCorrect", "mean"),
        subjects_attempted=("SubjectId", "nunique"),
        questions_attempted=("QuestionId", "nunique")
    )
    .reset_index()
)

print("Overall answer accuracy:")
print(f"{overall_accuracy:.2%}")

print("\nStudent profiles:", len(student_stats))

print("\nStudent statistics:")
display(
    student_stats.describe()
)

print("\nSample learner profiles:")
display(
    student_stats.head(10)
)
```

```python
# Preserve the dataset's interaction order for each learner

train["InteractionIndex"] = (
    train.groupby("UserId")
    .cumcount()
)

train["TotalUserInteractions"] = (
    train.groupby("UserId")["UserId"]
    .transform("size")
)

print("Interaction sequence created.")

display(
    train[
        [
            "UserId",
            "QuestionId",
            "SubjectId",
            "IsCorrect",
            "InteractionIndex",
            "TotalUserInteractions"
        ]
    ].head(10)
)
```

```python
# First 80% of each learner's interactions = history
# Final 20% = prediction/evaluation

train["IsHistory"] = (
    train["InteractionIndex"]
    < train["TotalUserInteractions"] * 0.8
)

history = train[
    train["IsHistory"]
].copy()

validation = train[
    ~train["IsHistory"]
].copy()

print("History interactions:", len(history))
print("Validation interactions:", len(validation))

print(
    f"History ratio: "
    f"{len(history) / len(train):.2%}"
)

print(
    f"Validation ratio: "
    f"{len(validation) / len(train):.2%}"
)

print(
    "\nValidation learners:",
    validation["UserId"].nunique()
)
```

```python
learner_history = (
    history.groupby("UserId")
    .agg(
        history_attempts=("IsCorrect", "count"),
        history_correct=("IsCorrect", "sum"),
        history_accuracy=("IsCorrect", "mean"),
        history_questions=("QuestionId", "nunique"),
        history_subjects=("SubjectId", "nunique")
    )
    .reset_index()
)

print("Learner history profiles:", len(learner_history))

display(
    learner_history.head(10)
)
```

```python
learner_subject_history = (
    history.groupby(
        ["UserId", "SubjectId"]
    )
    .agg(
        subject_attempts=("IsCorrect", "count"),
        subject_correct=("IsCorrect", "sum"),
        subject_accuracy=("IsCorrect", "mean")
    )
    .reset_index()
)

print(
    "Learner-subject profiles:",
    len(learner_subject_history)
)

display(
    learner_subject_history.head(10)
)
```

```python
question_history = (
    history.groupby("QuestionId")
    .agg(
        question_attempts=("IsCorrect", "count"),
        question_correct=("IsCorrect", "sum"),
        question_accuracy=("IsCorrect", "mean")
    )
    .reset_index()
)

print(
    "Question profiles:",
    len(question_history)
)

display(
    question_history.head(10)
)
```

```python
# Cell 21 — Fast subject mapping

# SubjectId already represents the question's subject hierarchy.
# We only need a question-level mapping, because the same
# QuestionId always belongs to the same subject hierarchy.

question_subjects = (
    train[["QuestionId", "SubjectId"]]
    .drop_duplicates("QuestionId")
    .copy()
)

print("Unique questions:", len(question_subjects))
print("Columns:", question_subjects.columns.tolist())

display(question_subjects.head(10))
```

```python
# Cell 22 — Create primary subject from the question-level mapping

def get_primary_subject(value):
    if isinstance(value, (list, tuple)):
        return value[0] if len(value) > 0 else -1

    if isinstance(value, str):
        try:
            value = value.strip("[]")
            parts = [x.strip() for x in value.split(",") if x.strip()]
            return int(parts[0]) if parts else -1
        except:
            return -1

    if pd.notna(value):
        return int(value)

    return -1


question_subjects["PrimarySubjectId"] = (
    question_subjects["SubjectId"]
    .apply(get_primary_subject)
)

print("Question mappings:", len(question_subjects))
print(
    "Unique primary subjects:",
    question_subjects["PrimarySubjectId"].nunique()
)

print("\nSample mappings:")
display(
    question_subjects[
        ["QuestionId", "SubjectId", "PrimarySubjectId"]
    ].head(10)
)
```

```python
# Cell 23 — Create model feature table

model_data = validation[
    [
        "UserId",
        "QuestionId",
        "SubjectId",
        "IsCorrect"
    ]
].copy()

# Attach learner-level historical performance
model_data = model_data.merge(
    learner_history,
    on="UserId",
    how="left"
)

print("Model data after learner features:")
print(model_data.shape)

print("\nColumns:")
print(model_data.columns.tolist())

display(model_data.head(10))
```

```python
# Cell 24 — Parse subject hierarchy and extract most specific subject

import ast

def get_specific_subject(value):
    if isinstance(value, (list, tuple)):
        return int(value[-1]) if len(value) > 0 else -1

    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)

            if isinstance(parsed, (list, tuple)):
                return int(parsed[-1]) if len(parsed) > 0 else -1

            return int(parsed)

        except (ValueError, SyntaxError, TypeError):
            return -1

    if pd.notna(value):
        return int(value)

    return -1


# Create the specific subject
question_subjects["PrimarySubjectId"] = (
    question_subjects["SubjectId"]
    .apply(get_specific_subject)
)

# Remove previous incorrect mapping
model_data = model_data.drop(
    columns=["PrimarySubjectId"],
    errors="ignore"
)

# Merge correct mapping
model_data = model_data.merge(
    question_subjects[
        ["QuestionId", "PrimarySubjectId"]
    ],
    on="QuestionId",
    how="left"
)

print("Model data:")
print(model_data.shape)

print("\nMissing specific subjects:")
print(model_data["PrimarySubjectId"].isna().sum())

print("\nInvalid subjects (-1):")
print((model_data["PrimarySubjectId"] == -1).sum())

print("\nUnique specific subjects:")
print(model_data["PrimarySubjectId"].nunique())

print("\nSample:")
display(
    model_data[
        [
            "QuestionId",
            "SubjectId",
            "PrimarySubjectId",
            "history_accuracy"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 27 — BUILD USER × SPECIFIC-SUBJECT FEATURES
# ============================================================

import re

# Work from the already-created learner subject history
subject_features = learner_subject_history.copy()

# Extract the final subject ID from the hierarchy
subject_features["SpecificSubjectId"] = (
    subject_features["SubjectId"]
    .str.extract(r',\s*(\d+)\]$')[0]
    .astype("int64")
)

print("Subject features before aggregation:")
print(subject_features.shape)

print("\nUnique specific subjects:")
print(subject_features["SpecificSubjectId"].nunique())

# Aggregate in case multiple subject-paths map to the same
# specific subject for the same learner
subject_features = (
    subject_features
    .groupby(
        ["UserId", "SpecificSubjectId"],
        as_index=False
    )
    .agg(
        subject_attempts=("subject_attempts", "sum"),
        subject_correct=("subject_correct", "sum")
    )
)

# Recalculate accuracy from aggregated counts
subject_features["subject_accuracy"] = (
    subject_features["subject_correct"]
    / subject_features["subject_attempts"].replace(0, 1)
)

print("\nSubject features after aggregation:")
print(subject_features.shape)

print("\nUnique learners:")
print(subject_features["UserId"].nunique())

print("\nUnique specific subjects:")
print(subject_features["SpecificSubjectId"].nunique())

print("\nSample:")
display(subject_features.head(10))
```

```python
# ============================================================
# CELL 28 — MERGE SPECIFIC SUBJECT FEATURES
# ============================================================

# Extract the specific subject ID from the validation question's
# subject hierarchy
model_data["SpecificSubjectId"] = (
    model_data["SubjectId"]
    .str.extract(r',\s*(\d+)\]$')[0]
    .astype("int64")
)

print("Model data specific subjects:")
print(model_data["SpecificSubjectId"].nunique())

# Remove the previously broken subject feature columns
old_subject_cols = [
    "PrimarySubjectId",
    "subject_attempts",
    "subject_correct",
    "subject_accuracy"
]

for col in old_subject_cols:
    if col in model_data.columns:
        model_data.drop(columns=col, inplace=True)


# Merge learner × specific-subject history
model_data = model_data.merge(
    subject_features[
        [
            "UserId",
            "SpecificSubjectId",
            "subject_attempts",
            "subject_correct",
            "subject_accuracy"
        ]
    ],
    on=["UserId", "SpecificSubjectId"],
    how="left"
)


# Learners with no previous history for this specific subject
missing_subject_history = (
    model_data["subject_attempts"].isna().sum()
)

print("\nMissing subject history:")
print(missing_subject_history)

print(
    "Missing percentage:",
    f"{missing_subject_history / len(model_data) * 100:.2f}%"
)


# Fill unseen learner-subject combinations
model_data["subject_attempts"] = (
    model_data["subject_attempts"].fillna(0)
)

model_data["subject_correct"] = (
    model_data["subject_correct"].fillna(0)
)

model_data["subject_accuracy"] = (
    model_data["subject_accuracy"].fillna(0.0)
)


print("\nModel data shape:")
print(model_data.shape)

print("\nColumns:")
print(model_data.columns.tolist())

print("\nSample:")
display(
    model_data[
        [
            "UserId",
            "QuestionId",
            "SpecificSubjectId",
            "history_attempts",
            "history_accuracy",
            "subject_attempts",
            "subject_correct",
            "subject_accuracy",
            "IsCorrect"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 29 — ADD QUESTION DIFFICULTY FEATURES
# ============================================================

question_features = question_history[
    [
        "QuestionId",
        "question_attempts",
        "question_correct",
        "question_accuracy"
    ]
].copy()

# Ensure matching key type
model_data["QuestionId"] = model_data["QuestionId"].astype("int64")
question_features["QuestionId"] = question_features["QuestionId"].astype("int64")

# Merge historical question performance
model_data = model_data.merge(
    question_features,
    on="QuestionId",
    how="left"
)

# Questions with no historical attempts
model_data["question_attempts"] = (
    model_data["question_attempts"].fillna(0)
)

model_data["question_correct"] = (
    model_data["question_correct"].fillna(0)
)

model_data["question_accuracy"] = (
    model_data["question_accuracy"].fillna(0.0)
)

print("=" * 60)
print("QUESTION FEATURES MERGED")
print("=" * 60)

print("Model data shape:", model_data.shape)

print("\nMissing question features:")
print(
    model_data[
        [
            "question_attempts",
            "question_correct",
            "question_accuracy"
        ]
    ].isna().sum()
)

print("\nSample:")
display(
    model_data[
        [
            "UserId",
            "QuestionId",
            "SpecificSubjectId",
            "history_attempts",
            "history_accuracy",
            "subject_attempts",
            "subject_accuracy",
            "question_attempts",
            "question_accuracy",
            "IsCorrect"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 30 — TRAIN LEARNER PERFORMANCE MODEL
# ============================================================

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score, roc_auc_score

print("=" * 60)
print("PREPARING CORRECTNESS MODEL")
print("=" * 60)

# Numerical behavioral features
FEATURES = [
    "IsCorrect",  # temporary target, removed below
    "history_attempts",
    "history_correct",
    "history_accuracy",
    "history_questions",
    "history_subjects",
    "subject_attempts",
    "subject_correct",
    "subject_accuracy",
    "question_attempts",
    "question_correct",
    "question_accuracy"
]

# Target
y = model_data["IsCorrect"].astype("int8")

# Remove target
feature_cols = [f for f in FEATURES if f != "IsCorrect"]

X = model_data[feature_cols].copy()

print("Features used:")
print(feature_cols)

print("\nTraining matrix:", X.shape)
print("Target distribution:")
print(y.value_counts(normalize=True))

# ------------------------------------------------------------
# TRAIN MODEL
# ------------------------------------------------------------

correctness_model = HistGradientBoostingClassifier(
    learning_rate=0.08,
    max_iter=150,
    max_leaf_nodes=31,
    min_samples_leaf=100,
    l2_regularization=1.0,
    random_state=42
)

print("\nTraining correctness model...")

correctness_model.fit(X, y)

print("\nCorrectness model trained successfully.")
```

```python
# ============================================================
# CELL 31 — EVALUATE CORRECTNESS MODEL
# ============================================================

print("=" * 60)
print("EVALUATING CORRECTNESS MODEL")
print("=" * 60)

# Predictions
correctness_pred = correctness_model.predict(X)
correctness_prob = correctness_model.predict_proba(X)[:, 1]

# Metrics
accuracy = accuracy_score(y, correctness_pred)
auc = roc_auc_score(y, correctness_prob)

print(f"\nAccuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"ROC-AUC:  {auc:.4f}")

print("\nPrediction distribution:")
print(
    pd.Series(correctness_pred)
    .value_counts()
    .sort_index()
)

print("\nProbability statistics:")
print(
    pd.Series(correctness_prob).describe()
)
```

```python
# ============================================================
# CELL 32 — SAVE CORRECTNESS MODEL
# ============================================================

import joblib
from pathlib import Path

# Create models directory
MODELS_DIR = Path("../models")
MODELS_DIR.mkdir(parents=True, exist_ok=True)

# Save model
correctness_model_path = MODELS_DIR / "learner_correctness_model.joblib"

joblib.dump(
    correctness_model,
    correctness_model_path
)

# Save feature names too
feature_names_path = MODELS_DIR / "learner_correctness_features.joblib"

joblib.dump(
    feature_cols,
    feature_names_path
)

print("=" * 60)
print("CORRECTNESS MODEL SAVED")
print("=" * 60)

print(f"Model:    {correctness_model_path.resolve()}")
print(f"Features: {feature_names_path.resolve()}")

print("\nFeatures saved:")
for i, feature in enumerate(feature_cols, 1):
    print(f"{i:2}. {feature}")
```
