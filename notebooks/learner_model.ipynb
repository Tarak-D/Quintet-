```python
import pandas as pd
import numpy as np
import joblib

from pathlib import Path
```

```python
ROOT = Path.cwd().parent

DATASETS = ROOT / "datasets"
INTERACTIONS = DATASETS / "eedi_interactions"

MODELS = ROOT / "models"
MODELS.mkdir(exist_ok=True)

print("Project root:")
print(ROOT)

print("\nEEDI interactions:")
print(INTERACTIONS)

print("\nModels:")
print(MODELS)
```

```python
print("EEDI interaction folders:\n")

for item in INTERACTIONS.iterdir():
    if item.is_dir():
        print(f"[FOLDER] {item.name}")
    else:
        print(f"[FILE]   {item.name}")
```

```python
TRAIN_DATA = INTERACTIONS / "train_data"

print("Training folder:")
print(TRAIN_DATA)

print("\nTraining files:\n")

for item in TRAIN_DATA.iterdir():
    if item.is_file():
        size_mb = item.stat().st_size / (1024 ** 2)
        print(f"{item.name:<45} {size_mb:>8.2f} MB")
    else:
        print(f"[FOLDER] {item.name}")
```

```python
METADATA = INTERACTIONS / "metadata"

print("Metadata folder:")
print(METADATA)

print("\nMetadata files:\n")

for item in METADATA.iterdir():
    if item.is_file():
        size_mb = item.stat().st_size / (1024 ** 2)
        print(f"{item.name:<50} {size_mb:>8.2f} MB")
    else:
        print(f"[FOLDER] {item.name}")
```

```python
TRAIN_FILE = TRAIN_DATA / "train_task_1_2.csv"

print("Loading:")
print(TRAIN_FILE)

train = pd.read_csv(TRAIN_FILE)

print("\nLoaded successfully.")
print("Shape:", train.shape)
```

```python
print("Columns:")
print(train.columns.tolist())

print("\nData types:")
print(train.dtypes)
```

```python
print("First 5 rows:")
display(train.head())

print("\nLast 5 rows:")
display(train.tail())
```

```python
print("Missing values:\n")

missing = train.isnull().sum()

print(missing)

print("\nMissing percentages:\n")

print(
    (missing / len(train) * 100)
    .round(3)
)
```

```python
print("Unique values:\n")

for column in train.columns:
    print(
        f"{column:<25} "
        f"{train[column].nunique():,}"
    )
```

```python
# Reduce memory usage

for col in train.columns:
    if train[col].dtype == "int64":
        train[col] = pd.to_numeric(
            train[col],
            downcast="integer"
        )

print("Memory usage after optimization:")
print(
    f"{train.memory_usage(deep=True).sum() / (1024**2):.2f} MB"
)

print("\nData types:")
print(train.dtypes)
```

```python
QUESTION_META = (
    METADATA / "question_metadata_task_1_2.csv"
)

question_meta = pd.read_csv(
    QUESTION_META
)

print("Question metadata shape:", question_meta.shape)

print("\nColumns:")
print(question_meta.columns.tolist())

display(question_meta.head())
```

```python
train_questions = set(
    train["QuestionId"].unique()
)

metadata_questions = set(
    question_meta["QuestionId"].unique()
)

missing_questions = (
    train_questions - metadata_questions
)

print("Questions in training:", len(train_questions))
print("Questions in metadata:", len(metadata_questions))
print("Training questions missing metadata:", len(missing_questions))

if len(missing_questions) > 0:
    print("\nSample missing IDs:")
    print(list(missing_questions)[:20])
else:
    print("\nAll training questions have subject metadata.")
```

```python
train = train.merge(
    question_meta[
        ["QuestionId", "SubjectId"]
    ],
    on="QuestionId",
    how="left"
)

print("Shape after merge:", train.shape)

print("\nMissing SubjectIds:")
print(train["SubjectId"].isna().sum())

print("\nSample:")
display(train.head())
```

```python
overall_accuracy = train["IsCorrect"].mean()

student_stats = (
    train.groupby("UserId")
    .agg(
        total_attempts=("IsCorrect", "count"),
        correct_answers=("IsCorrect", "sum"),
        accuracy=("IsCorrect", "mean"),
        subjects_attempted=("SubjectId", "nunique"),
        questions_attempted=("QuestionId", "nunique")
    )
    .reset_index()
)

print("Overall answer accuracy:")
print(f"{overall_accuracy:.2%}")

print("\nStudent profiles:", len(student_stats))

print("\nStudent statistics:")
display(
    student_stats.describe()
)

print("\nSample learner profiles:")
display(
    student_stats.head(10)
)
```

```python
# Preserve the dataset's interaction order for each learner

train["InteractionIndex"] = (
    train.groupby("UserId")
    .cumcount()
)

train["TotalUserInteractions"] = (
    train.groupby("UserId")["UserId"]
    .transform("size")
)

print("Interaction sequence created.")

display(
    train[
        [
            "UserId",
            "QuestionId",
            "SubjectId",
            "IsCorrect",
            "InteractionIndex",
            "TotalUserInteractions"
        ]
    ].head(10)
)
```

```python
# First 80% of each learner's interactions = history
# Final 20% = prediction/evaluation

train["IsHistory"] = (
    train["InteractionIndex"]
    < train["TotalUserInteractions"] * 0.8
)

history = train[
    train["IsHistory"]
].copy()

validation = train[
    ~train["IsHistory"]
].copy()

print("History interactions:", len(history))
print("Validation interactions:", len(validation))

print(
    f"History ratio: "
    f"{len(history) / len(train):.2%}"
)

print(
    f"Validation ratio: "
    f"{len(validation) / len(train):.2%}"
)

print(
    "\nValidation learners:",
    validation["UserId"].nunique()
)
```

```python
learner_history = (
    history.groupby("UserId")
    .agg(
        history_attempts=("IsCorrect", "count"),
        history_correct=("IsCorrect", "sum"),
        history_accuracy=("IsCorrect", "mean"),
        history_questions=("QuestionId", "nunique"),
        history_subjects=("SubjectId", "nunique")
    )
    .reset_index()
)

print("Learner history profiles:", len(learner_history))

display(
    learner_history.head(10)
)
```

```python
learner_subject_history = (
    history.groupby(
        ["UserId", "SubjectId"]
    )
    .agg(
        subject_attempts=("IsCorrect", "count"),
        subject_correct=("IsCorrect", "sum"),
        subject_accuracy=("IsCorrect", "mean")
    )
    .reset_index()
)

print(
    "Learner-subject profiles:",
    len(learner_subject_history)
)

display(
    learner_subject_history.head(10)
)
```

```python
question_history = (
    history.groupby("QuestionId")
    .agg(
        question_attempts=("IsCorrect", "count"),
        question_correct=("IsCorrect", "sum"),
        question_accuracy=("IsCorrect", "mean")
    )
    .reset_index()
)

print(
    "Question profiles:",
    len(question_history)
)

display(
    question_history.head(10)
)
```

```python
# Cell 21 — Fast subject mapping

# SubjectId already represents the question's subject hierarchy.
# We only need a question-level mapping, because the same
# QuestionId always belongs to the same subject hierarchy.

question_subjects = (
    train[["QuestionId", "SubjectId"]]
    .drop_duplicates("QuestionId")
    .copy()
)

print("Unique questions:", len(question_subjects))
print("Columns:", question_subjects.columns.tolist())

display(question_subjects.head(10))
```

```python
# Primary subject = first subject in the EEDI hierarchy

train["PrimarySubjectId"] = train["SubjectList"].apply(
    lambda x: x[0] if len(x) > 0 else -1
)

print(
    "Unique primary subjects:",
    train["PrimarySubjectId"].nunique()
)

display(
    train[
        ["QuestionId", "SubjectList", "PrimarySubjectId"]
    ].head(10)
)
```

```python
# Merge historical learner statistics into validation interactions

model_data = validation[
    [
        "UserId",
        "QuestionId",
        "SubjectList",
        "PrimarySubjectId",
        "IsCorrect"
    ]
].copy()

model_data = model_data.merge(
    learner_history,
    on="UserId",
    how="left"
)

print("After learner merge:")
print(model_data.shape)

display(model_data.head())
```

```python
# Build the learner-subject key using the primary subject

subject_features = learner_subject_history.copy()

subject_features = subject_features.rename(
    columns={
        "SubjectId": "PrimarySubjectId"
    }
)

model_data = model_data.merge(
    subject_features,
    on=["UserId", "PrimarySubjectId"],
    how="left"
)

print("After subject mastery merge:")
print(model_data.shape)

display(
    model_data[
        [
            "UserId",
            "QuestionId",
            "PrimarySubjectId",
            "history_accuracy",
            "subject_attempts",
            "subject_accuracy"
        ]
    ].head(10)
)
```

```python
model_data = model_data.merge(
    question_history,
    on="QuestionId",
    how="left"
)

# Fill cases where the learner has never attempted
# the subject before.

model_data["subject_attempts"] = (
    model_data["subject_attempts"].fillna(0)
)

model_data["subject_correct"] = (
    model_data["subject_correct"].fillna(0)
)

model_data["subject_accuracy"] = (
    model_data["subject_accuracy"].fillna(
        model_data["history_accuracy"]
    )
)

# Fill questions with no historical observations
model_data["question_attempts"] = (
    model_data["question_attempts"].fillna(0)
)

model_data["question_correct"] = (
    model_data["question_correct"].fillna(0)
)

model_data["question_accuracy"] = (
    model_data["question_accuracy"].fillna(
        model_data["history_accuracy"]
    )
)

print("Final feature table prepared.")

print("Shape:", model_data.shape)

print("\nMissing values:")
print(
    model_data[
        [
            "history_accuracy",
            "subject_accuracy",
            "question_accuracy"
        ]
    ].isna().sum()
)

display(model_data.head(10))
```
