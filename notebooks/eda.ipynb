```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
```

```python
ROOT = Path.cwd().parent

DATASETS = ROOT / "datasets"

EEDI_INTERACTIONS = DATASETS / "eedi_interactions"
EEDI_MISCONCEPTIONS = DATASETS / "eedi_misconceptions"

print("Project root:", ROOT)
print("Datasets:", DATASETS)
```

```python
subjects = pd.read_csv(
    metadata_path / "subject_metadata.csv"
)

print(subjects.head(10))
print("\nLevels:")
print(subjects["Level"].value_counts().sort_index())

print("\nMissing values:")
print(subjects.isnull().sum())
```

```python
miscon_path = EEDI_MISCONCEPTIONS

train_mis = pd.read_csv(miscon_path / "train.csv")
mapping = pd.read_csv(miscon_path / "misconception_mapping.csv")

print("MISCONCEPTION TRAIN")
print("Shape:", train_mis.shape)
print("\nColumns:")
print(train_mis.columns.tolist())

print("\nFirst 3 rows:")
display(train_mis.head(3))

print("\nMISCONCEPTION MAPPING")
print("Shape:", mapping.shape)
print(mapping.head(5))
```

```python
miscon_cols = [
    "MisconceptionAId",
    "MisconceptionBId",
    "MisconceptionCId",
    "MisconceptionDId"
]

print("Misconception labels per option:\n")

for col in miscon_cols:
    count = train_mis[col].notna().sum()
    print(f"{col}: {count:,} labeled")

print("\nTotal labeled distractors:",
      train_mis[miscon_cols].notna().sum().sum())

print("\nUnique misconception IDs:",
      pd.unique(train_mis[miscon_cols].values.ravel()).shape[0] - 1)
```

```python
print("Columns:")
print(mapping.columns.tolist())

print("\nShape:", mapping.shape)

display(mapping.head(10))

print("\nMissing values:")
print(mapping.isnull().sum())
```

```python
misconception_ids = pd.unique(
    train_mis[miscon_cols].values.ravel()
)

misconception_ids = misconception_ids[
    pd.notna(misconception_ids)
].astype(int)

print("Unique misconception IDs:", len(np.unique(misconception_ids)))
print("Minimum ID:", misconception_ids.min())
print("Maximum ID:", misconception_ids.max())

print("\nMost frequent misconceptions:")
print(
    pd.Series(misconception_ids)
    .value_counts()
    .head(15)
)
```

```python
question_meta_12 = pd.read_csv(
    metadata_path / "question_metadata_task_1_2.csv"
)

print("Dataset 1 QuestionId range:")
print(
    question_meta_12["QuestionId"].min(),
    "→",
    question_meta_12["QuestionId"].max()
)

print("\nDataset 2 QuestionId range:")
print(
    train_mis["QuestionId"].min(),
    "→",
    train_mis["QuestionId"].max()
)

print("\nDataset 1 unique QuestionIds:",
      question_meta_12["QuestionId"].nunique())

print("Dataset 2 unique QuestionIds:",
      train_mis["QuestionId"].nunique())
```

```python
ids_1 = set(question_meta_12["QuestionId"])
ids_2 = set(train_mis["QuestionId"])

overlap = ids_1 & ids_2

print("Question IDs in Dataset 1:", len(ids_1))
print("Question IDs in Dataset 2:", len(ids_2))
print("Directly overlapping IDs:", len(overlap))

print("\nPercentage of Dataset 2 IDs overlapping:",
      f"{len(overlap) / len(ids_2) * 100:.2f}%")

print("\nFirst 20 overlapping IDs:")
print(sorted(overlap)[:20])
```

```python
fraction_mis = train_mis[
    train_mis["SubjectId"] == 39
].copy()

print("Fraction misconception questions:",
      len(fraction_mis))

print("\nUnique constructs:",
      fraction_mis["ConstructId"].nunique())

print("\nConstructs:")
display(
    fraction_mis[
        ["ConstructId", "ConstructName"]
    ].drop_duplicates()
    .sort_values("ConstructId")
)
```

```python
subject_dist = (
    train_mis.groupby(["SubjectId", "SubjectName"])
    .size()
    .sort_values(ascending=False)
)

print("Unique subjects:", train_mis["SubjectId"].nunique())

display(subject_dist.head(30))
```

```python
fraction_subjects = train_mis[
    train_mis["SubjectName"].str.contains(
        "fraction", case=False, na=False
    )
][["SubjectId", "SubjectName"]].drop_duplicates()

print("Fraction-related subjects:")
display(fraction_subjects.sort_values("SubjectId"))
```

```python
first_questions = question_meta_12[
    question_meta_12["QuestionId"].isin(
        train_mis["QuestionId"]
    )
].copy()

print("Matched QuestionIds:", len(first_questions))

print("\nDataset 1 subjects represented:")
display(
    first_questions["SubjectId"]
    .value_counts()
    .head(20)
)
```

```python
comparison = train_mis.merge(
    question_meta_12,
    on="QuestionId",
    how="left",
    suffixes=("_miscon", "_eedi")
)

print("Rows:", len(comparison))
print("Missing EEDI SubjectId:",
      comparison["SubjectId_eedi"].isna().sum())

display(
    comparison[
        [
            "QuestionId",
            "SubjectId_miscon",
            "SubjectName",
            "SubjectId_eedi"
        ]
    ].head(20)
)
```
